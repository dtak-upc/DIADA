```python
"""
Experiment 3 analysis: DIADA+ARDA sweep results (results_final_*.csv).

Reads the frozen paper-original results in this same folder. Plotting
logic lives in diada.experiments.results_plotting (shared with Experiment
4's notebook -- see that module's docstring for what's shared and why).
"""
import pandas as pd

from diada.experiments import results_plotting as rp

FILES = {
    "classification": "results_final_classification.csv",
    "regression": "results_final_regression.csv",
}

variants = {
    "noisy_1cf": {"suffix": "_un_dirty", "color": "lightcoral", "label": "UN pre-comp.", "linestyle": "-", "marker": "o"},
    "cleaned_1cf": {"suffix": "_un_clean", "color": "lightgreen", "label": "UN post-comp.", "linestyle": "-", "marker": "o"},
    "noisy_2cf": {"suffix": "_mn_dirty", "color": "firebrick", "label": "MN pre-comp.", "linestyle": "--", "marker": "o"},
    "cleaned_2cf": {"suffix": "_mn_clean", "color": "forestgreen", "label": "MN post-comp.", "linestyle": "--", "marker": "o"},
    "normal": {"suffix": "", "color": "grey", "label": "Base", "linestyle": ":", "marker": "o"},
}

MAX_FEATURES_PER_BENCHMARK = {
    "bank": 15, "default_payment": 23, "jannis": 40, "miniboone": 40, "appliances": 27,
    "avocado_sales": 12, "diamonds": 8, "house_sales": 16, "nasa": 21, "pol": 40,
    "superconductivity": 40, "covertype": 12, "drive_diagnosis": 40, "dry_beans": 16,
    "mice_protein": 40, "pendigits": 16, "students": 35,
}

df_class = pd.read_csv(FILES["classification"])
df_reg = pd.read_csv(FILES["regression"])

rp.plot_dataset(
    df_class, task_type="classification", metric_col="test_accuracy", metric_label="Accuracy",
    variants=variants, max_features_per_benchmark=MAX_FEATURES_PER_BENCHMARK,
    classification_priority_dataset="mice_protein",
    classification_excluded_datasets=["pendigits", "covertype", "drive_diagnosis"],
)
rp.plot_dataset(
    df_reg, task_type="regression", metric_col="test_rmse", metric_label="RMSE",
    variants=variants, max_features_per_benchmark=MAX_FEATURES_PER_BENCHMARK,
)
```

```python
# Noise-type breakdown (uncleaned "_dirty" variants only), at the largest
# num_new_features tested for each.
rp.print_noise_table(df_class)
rp.print_noise_table(df_reg)
```

```python
# Augmentation-time comparison across variants.
#
# NOTE: the original results.ipynb read "results_final_classification_2.csv"
# / "results_final_regression_2.csv" here -- files that don't exist
# anywhere in this repo (neither committed nor produced by any script).
# Fixed to read the real results_final_*.csv files already loaded above
# instead of silently carrying the stale reference forward.
aug_df = pd.concat(
    [df_class[["benchmark", "augmentation_time"]], df_reg[["benchmark", "augmentation_time"]]]
).drop_duplicates().reset_index(drop=True)

aug_variants = {k: v for k, v in variants.items() if v["suffix"]}  # drop the no-suffix "Base" entry
rp.plot_augmentation_time_bars(aug_df, variants=aug_variants)
```
