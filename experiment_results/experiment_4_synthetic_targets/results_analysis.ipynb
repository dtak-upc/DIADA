```python
"""
Experiment 4 analysis: synthetic-target ablation results
(results_2targets_*.csv).

Tests whether ARDA can be fooled into selecting features that predict a
synthetic decoy target (correlated with a subset of the real features)
rather than the real one. Reads the frozen paper-original results in this
same folder. Plotting logic lives in diada.experiments.results_plotting
(shared with Experiment 3's notebook -- see that module's docstring).

No live-rerun script exists yet for this experiment (see EXPERIMENTS.md) --
this notebook only analyzes the frozen results.
"""
import pandas as pd

from diada.experiments import results_plotting as rp

FILES = {
    "classification": "results_2targets_classification.csv",
    "regression": "results_2targets_regression.csv",
}

variants = {
    "target_1_noise": {"suffix": "_synthetic_target_1_noise", "color": "lightcoral", "label": "Target 1 pre-comp.", "linestyle": "-", "marker": "o"},
    "target_1_clean": {"suffix": "_synthetic_target_1_clean", "color": "lightgreen", "label": "Target 1 post-comp.", "linestyle": "-", "marker": "o"},
    "target_1_base": {"suffix": "_synthetic_target_1_base", "color": "grey", "label": "Target 1 base", "linestyle": "-", "marker": "o"},
    "target_2_noise": {"suffix": "_synthetic_target_2_noise", "color": "firebrick", "label": "Target 2 pre-comp.", "linestyle": "--", "marker": "o"},
    "target_2_clean": {"suffix": "_synthetic_target_2_clean", "color": "forestgreen", "label": "Target 2 post-comp.", "linestyle": "--", "marker": "o"},
    "target_2_base": {"suffix": "_synthetic_target_2_base", "color": "black", "label": "Target 2 base", "linestyle": "--", "marker": "o"},
}

MAX_FEATURES_PER_BENCHMARK = {
    "bank": 15, "default_payment": 23, "jannis": 40, "miniboone": 40, "appliances": 27,
    "avocado_sales": 12, "diamonds": 8, "house_sales": 16, "nasa": 21, "pol": 40,
    "superconductivity": 40, "covertype": 12, "drive_diagnosis": 40, "dry_beans": 16,
    "mice_protein": 40, "pendigits": 16, "students": 35,
}

df_class = pd.read_csv(FILES["classification"])
df_reg = pd.read_csv(FILES["regression"])

# Only the legend text size differs from results_plotting's default style
# (9 here vs. 11 for Experiment 3 -- more entries in this experiment's
# legend, so the original notebook shrank it to fit).
style = {**rp.DEFAULT_STYLE, "legend_size": 9}

rp.plot_dataset(
    df_class, task_type="classification", metric_col="test_accuracy", metric_label="Accuracy",
    variants=variants, max_features_per_benchmark=MAX_FEATURES_PER_BENCHMARK, style=style,
    classification_priority_dataset="miniboone",
    classification_excluded_datasets=["pendigits", "covertype", "drive_diagnosis"],
)
rp.plot_dataset(
    df_reg, task_type="regression", metric_col="test_rmse", metric_label="RMSE",
    variants=variants, max_features_per_benchmark=MAX_FEATURES_PER_BENCHMARK, style=style,
)
```
